# 03-3 조건문과 논리 실습

비교, 멤버십, 동일성, 논리 연산자, 우선순위, 단락 평가, truthy/falsey, 결정표를 한 이벤트의 분류 로직에 적용합니다.

## 1. 조건식과 비교

실행 전에 각 결과를 예상하세요.

```python
port = 443
action = "DENY"

print(port == 443)
print(port != 80)
print(1 <= port <= 65535)
print(action in {"ALLOW", "DENY"})
print(action not in {"BLOCK", "DROP"})
```

## 2. 경계값 검증

최솟값, 최댓값, 경계 바로 밖을 함께 확인합니다.

```python
print(1 <= 0 <= 65535)
print(1 <= 1 <= 65535)
print(1 <= 65535 <= 65535)
print(1 <= 65536 <= 65535)
```

## 3. ==와 is

값 비교와 객체 동일성 비교의 차이를 확인합니다.

```python
a = ["DENY"]
b = ["DENY"]
country = None

print(a == b)
print(a is b)
print(country is None)
print(country is not None)
```

## 4. and, or, not 진리표

네 조합의 결과를 실행 전에 적어 보세요.

```python
print(False and False, False or False, not False)
print(False and True,  False or True,  not False)
print(True and False,  True or False,  not True)
print(True and True,   True or True,   not True)
```

## 5. 우선순위와 괄호

`and`는 `or`보다 먼저 평가됩니다.

```python
A = True
B = False
C = False

print(A or B and C)
print(A or (B and C))
print((A or B) and C)
```

## 6. 단락 평가

앞 조건이 뒤 연산을 안전하게 실행할 수 있는지 결정합니다.

```python
user = None
is_admin = user is not None and user.startswith("admin")
print(is_admin)

parts = ["DENY", "198.51.100.9"]
has_port = len(parts) >= 3 and parts[2].isdigit()
print(has_port)
```

## 7. and와 or의 실제 반환값

논리 연산 결과가 항상 `bool`인 것은 아닙니다.

```python
print("" or "UNKNOWN")
print(None or "UNKNOWN")
print("DENY" or "UNKNOWN")
print("DENY" and 443)
print("" and 443)

configured_timeout = 0
wrong_timeout = configured_timeout or 30
correct_timeout = 30 if configured_timeout is None else configured_timeout
print(wrong_timeout, correct_timeout)
```

## 8. truthy/falsey와 업무 의미

`None`, 0, 빈 문자열은 모두 falsey지만 같은 상태는 아닙니다.

```python
count = 0

if count is None:
    count_status = "NOT_COUNTED"
elif count == 0:
    count_status = "COUNTED_ZERO"
else:
    count_status = "COUNTED_POSITIVE"

print(count_status)
```

## 9. 드모르간 법칙, any, all

복합 조건의 부정과 여러 검증 결과의 결합을 확인합니다.

```python
A = True
B = False

demorgan_and = (not (A and B)) == ((not A) or (not B))
demorgan_or = (not (A or B)) == ((not A) and (not B))
print(demorgan_and)
print(demorgan_or)
assert demorgan_and is True
assert demorgan_or is True

validation_results = [True, True, False]
print(any(validation_results))
print(all(validation_results))
print(any([]))
print(all([]))
```

## 10. 독립된 if와 elif

독립된 `if`는 여러 규칙이 동시에 실행될 수 있고, `if`·`elif`는 하나의 등급만 선택합니다.

```python
failed_count = 7

if failed_count >= 3:
    print("독립 if: 경고 조건")
if failed_count >= 5:
    print("독립 if: 심각 조건")

if failed_count >= 5:
    level = "CRITICAL"
elif failed_count >= 3:
    level = "WARNING"
else:
    level = "NORMAL"

print("elif 분류:", level)
```

## 11. 미니 실습 — 단일 이벤트 분류

유효성 검증과 위험 분류를 별도 bool 변수로 만든 뒤 우선순위에 따라 하나의 등급을 선택합니다.

```python
event = {
    "action": "DENY",
    "port": 443,
    "path": "/admin",
    "failed_count": 4,
    "country": None,
}

action = event["action"]
port = event["port"]
path = event["path"]
failed_count = event["failed_count"]

is_known_action = action in {"ALLOW", "DENY"}
is_valid_port = 1 <= port <= 65535
is_valid_input = is_known_action and is_valid_port

is_deny = action == "DENY"
is_sensitive_path = path in {"/admin", "/login"}
has_repeated_failures = failed_count >= 3
is_country_unknown = event.get("country") is None

if not is_valid_input:
    level = "INVALID"
elif is_deny and is_sensitive_path and has_repeated_failures:
    level = "CRITICAL"
elif is_deny or is_sensitive_path:
    level = "REVIEW"
else:
    level = "NORMAL"

print(level)
```

### 자기점검

아래 셀이 오류 없이 끝나면 필수 실습을 완료한 것입니다.

```python
assert is_known_action is True
assert is_valid_port is True
assert is_valid_input is True
assert is_deny is True
assert is_sensitive_path is True
assert has_repeated_failures is True
assert is_country_unknown is True
assert level == "CRITICAL"

print("모든 자기점검을 통과했습니다.")
```
